# 📘 Resumen de chats con Cadenas de Markov

Este notebook implementa un modelo de Markov para **resumir conversaciones** y transformarlas en un formato **narrativo fluido**.

## Pasos:
1. Entrenar un modelo de Markov con el texto de un chat.
2. Generar un resumen crudo.
3. Limpiar repeticiones.
4. Convertir el resultado en narrativa.


```python
import random
import re
from collections import defaultdict

# ================================
# 1. Modelo de Markov simple
# ================================
def entrenar_markov(texto, n=2):
    """
    Entrena un modelo de Markov a partir de un texto dado.
    - texto: corpus en formato string.
    - n: tamaño del n-grama (2 = bigrama, 3 = trigrama).
    Retorna un diccionario con el modelo entrenado.
    """
    palabras = texto.split()
    modelo = defaultdict(list)
    for i in range(len(palabras) - n):
        clave = tuple(palabras[i:i+n])
        modelo[clave].append(palabras[i+n])
    return modelo

def generar_texto(modelo, longitud=80):
    """
    Genera un texto a partir de un modelo de Markov entrenado.
    - modelo: diccionario entrenado.
    - longitud: cantidad de palabras a generar.
    """
    inicio = random.choice(list(modelo.keys()))
    resultado = list(inicio)
    for _ in range(longitud):
        clave = tuple(resultado[-len(inicio):])
        if clave in modelo:
            siguiente = random.choice(modelo[clave])
            resultado.append(siguiente)
        else:
            break
    return " ".join(resultado)

# ================================
# 2. Limpieza de repeticiones
# ================================
def limpiar_repeticiones(texto):
    """
    Elimina repeticiones consecutivas de palabras.
    Ejemplo: "hola hola" -> "hola"
    """
    return re.sub(r'\b(\w+)( \1\b)+', r'\1', texto, flags=re.IGNORECASE)

# ================================
# 3. Conversión a narrativo
# ================================
def convertir_a_narrativo(texto: str) -> str:
    """
    Convierte diálogos tipo 'Juan: ... Carla: ...'
    en narrativa fluida.
    """
    partes = texto.split(":")
    narrativo = []
    
    for i in range(len(partes) - 1):
        nombre = partes[i].strip().split()[-1]
        contenido = partes[i+1].strip()
        
        if not contenido:
            continue
        
        if i == 0:
            narrativo.append(f"{nombre} comentó que {contenido[0].lower() + contenido[1:]}")
        else:
            narrativo.append(f"{nombre} añadió que {contenido[0].lower() + contenido[1:]}")
    
    return " ".join(narrativo)

# ================================
# 4. Demo con un ejemplo
# ================================
corpus = """
IA: Hola, ¿qué tal tu día?
Niño: Bien, estaba mirando un libro de planetas.
IA: ¡Qué interesante! ¿Tienes un planeta favorito?
Niño: Sí, Saturno, porque tiene anillos.
IA: Los anillos de Saturno están hechos de hielo y rocas pequeñas. ¿Sabías eso?
Niño: No, pensaba que eran como aros de metal.
IA: Jajaja, mucha gente lo imagina así. En realidad son millones de partículas girando.
Niño: ¿Y se pueden ver con un telescopio normal?
IA: Sí, con un telescopio pequeño puedes ver los anillos como una banda alrededor del planeta.
Niño: ¡Yo quiero uno!
IA: Sería un gran regalo. ¿Quieres que te cuente un dato de Saturno?
Niño: Sí, dime.
IA: Es tan grande que dentro de Saturno cabrían más de 700 Tierras.
Niño: ¡En serio! 😲
IA: Sí, y además flota en el agua, porque es menos denso que ella.
Niño: ¿O sea que Saturno es como un globo gigante?
IA: Exacto, es un planeta gaseoso.
Niño: Entonces no se puede parar uno ahí.
IA: Correcto, no hay superficie sólida.
Niño: Yo pensaba que los astronautas podían bajar.
IA: Solo en planetas rocosos, como Marte o la Luna.
Niño: Ahh, ahora entiendo.
IA: ¿Quieres que te cuente un secreto de los planetas?
Niño: ¡Sí!
IA: Todos los planetas giran alrededor del Sol en la misma dirección.
Niño: ¿Como una fila?
IA: Más bien como una pista de carreras, cada uno en su carril.
Niño: ¡Eso suena chévere!
"""

# Entrenamiento
modelo = entrenar_markov(corpus)

# Resumen crudo
resumen_crudo = generar_texto(modelo, longitud=60)
# Narrativo
resumen_final = limpiar_repeticiones(convertir_a_narrativo(generar_texto(modelo, longitud=60)))

print("\n=== Resumen narrativo ===")
print(resumen_final)
```

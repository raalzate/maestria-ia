# Ejemplo simple de Transformer con PyTorch

```

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.nn import Transformer
```

```

# Creamos un batch pequeño de datos de juguete
# Suponemos que cada "secuencia" tiene 5 tokens y cada token es un embedding de 16 dimensiones
src = torch.rand((5, 10, 16))  # (secuencia, batch, embedding)
tgt = torch.rand((5, 10, 16))
```

```

# Definimos un modelo Transformer simple
transformer = Transformer(d_model=16, nhead=2, num_encoder_layers=2, num_decoder_layers=2)

# Pasamos los datos por el modelo
output = transformer(src, tgt)

print("Forma de la salida:", output.shape)
```


En este ejemplo:
- Creamos datos de juguete con 5 tokens por secuencia, batch de tamaño 10, y dimensión de embedding 16.
- Definimos un `Transformer` básico con 2 capas de encoder y decoder, y 2 *heads* de atención.
- La salida mantiene la forma `(longitud_secuencia, batch, d_model)`, en este caso `(5, 10, 16)`.

Este es un **ejemplo mínimo**. En práctica:
- Se usan embeddings entrenables para convertir texto a vectores.
- Se añaden *positional encodings*.
- Se conecta a una capa final de clasificación o predicción (por ejemplo, vocabulario de tokens).

# Introducción a PyTorch (Explicado paso a paso para principiantes)


Este notebook es una **guía sencilla** para entender qué es PyTorch y cómo usarlo.

PyTorch es una librería de **inteligencia artificial** que nos permite:
- Manejar tensores (como matrices o arrays de NumPy, pero con soporte para GPU).
- Construir y entrenar **redes neuronales**.
- Hacer experimentos de aprendizaje automático de manera flexible.


```

import torch
import torch.nn as nn
import torch.optim as optim
```

## 1. Tensores en PyTorch

```

# Crear un tensor (matriz)
x = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
print("Tensor x:\n", x)

# Operaciones como en NumPy
y = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)
print("Suma:\n", x + y)
print("Multiplicación elemento a elemento:\n", x * y)

# Se pueden mover a GPU (si está disponible)
if torch.cuda.is_available():
    x = x.to("cuda")
    print("Tensor en GPU:", x)
```

## 2. Autograd: Gradientes automáticos

```

# PyTorch calcula derivadas automáticamente

a = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(3.0, requires_grad=True)

# Definimos una función: f = a^2 + b^2
f = a**2 + b**2
f.backward()  # Calcula derivadas

print("df/da =", a.grad.item())
print("df/db =", b.grad.item())
```

## 3. Construir una red neuronal simple

```

# Creamos datos ficticios
X = torch.randn(10, 3)  # 10 muestras, 3 características
y = torch.randn(10, 1)  # 10 etiquetas

# Definimos un modelo de red neuronal muy simple (una sola capa lineal)
model = nn.Linear(3, 1)

# Definimos la función de pérdida (error cuadrático medio)
criterion = nn.MSELoss()

# Definimos un optimizador (descenso de gradiente estocástico)
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Entrenamos por algunas épocas
for epoch in range(100):
    # Forward: predicción
    y_pred = model(X)
    loss = criterion(y_pred, y)

    # Backward: cálculo de gradientes
    optimizer.zero_grad()
    loss.backward()

    # Actualizamos parámetros
    optimizer.step()

print("Pérdida final:", loss.item())
```


## 🎯 Conclusión

En este notebook aprendimos:
1. Qué son los **tensores** en PyTorch.
2. Cómo funciona el **autograd** para calcular derivadas automáticamente.
3. Cómo construir y entrenar una **red neuronal muy simple**.

👉 Este es solo el comienzo: con PyTorch se pueden crear modelos mucho más avanzados como CNNs, RNNs o Transformers.
